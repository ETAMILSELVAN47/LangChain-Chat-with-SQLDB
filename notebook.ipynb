```python
import sqlite3
```

```python
# create a connection
connection=sqlite3.connect('student.db')
```

```python
# create a cursor object to create table,insert records
cursor=connection.cursor()
```

```python
# create table
table_info="""
create table STUDENT (name VARCHAR(25),class varchar(25),section varchar(25),marks int)
"""
```

```python
cursor.execute(table_info)
```

```python
# Insert records
cursor.execute('''Insert into STUDENT values ('Krish','Data science','A',90)''')
cursor.execute('''Insert into STUDENT values ('Naik','Data science','B',80)''')
cursor.execute('''Insert into STUDENT values ('John','Data science','A',70)''')
cursor.execute('''Insert into STUDENT values ('Mark','Data science','C',85)''')
```

```python
# display the records
data=cursor.execute('''SELECT * FROM STUDENT''')
dataset=list()
for row in data:
    dataset.append(row)
```

```python
import pandas as pd
pd.DataFrame(data=dataset,columns='name class section marks'.split())
```
